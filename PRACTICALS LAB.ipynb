```python
#Handling missing data using pandas
import pandas as pd
import numpy as np

#Creating dataframe with missing values
Data={
	"Name":["lakshya","manasa","sharanya","kaavya"],
    "Marks":[95,np.nan,89,np.nan],
    "City":["Chennai","delhi",np.nan,"Mumbai"]
}

df=pd.DataFrame(Data)

#  Display the original DataFrame
print("Original DataFrame:")
print(df)

# Check missing values using isnull()
print("\nMissing values (True/False):")
print(df.isnull())

# Count missing values in each column
print("\nCount of missing values per column:")
print(df.isnull().sum())

# Fill all missing values with a default value using fillna()
df_filled_default = df.fillna("Unknown")
print("\nAfter fillna() with default value 'Unknown':")
print(df_filled_default)

#  Fill missing values in Marks column with column mean using fillna()
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
print("\nAfter fillna() - Marks column filled with mean:")
print(df)

# Forward fill using fillna()
df_ffill = df.ffill
print("\nAfter fill - Forward Fill:")
print(df_ffill)

# Backward fill using fillna()
df_bfill = df.bfill
print("\nAfter fill - Backward Fill:")
print(df_bfill)

#  Remove rows with missing values using dropna()
df_dropped = df.dropna()
print("\nAfter dropna() - rows with missing values removed:")
print(df_dropped)

#  Replace a specified value using replace()
df_replaced = df.replace("Chennai", "Coimbatore")
print("\nAfter replace() - 'Chennai' replaced with 'Coimbatore':")
print(df_replaced)

#  Display DataFrame info
print("\nDataFrame Info:")
df.info()

```

```python
# Merge and Join Dataframes using Pandas

import pandas as pd

# ---------------- CONCATENATE OPERATIONS ----------------
# (All columns name should be same)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])

print("\nDataFrame 1:")
print(df1)

# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [201, 202, 203, 205],
    "Marks": [85, 78, 80, 65]
}, index=[4, 5, 6, 7])

print("\nDataFrame 2:")
print(df2)

# Concatenate DataFrames
frames = [df1, df2]
result = pd.concat(frames)

print("Concatenate DataFrames:")
print(result)

#---------------MERGE OPERATIONS-------------(atleastone column should be same,others can be different)

#Creating First Dataframes
df1=pd.DataFrame({
        "ID":[101,102,103,104],
        "Names":["lucky","anan","kaavs","padma"]
} )

# Merge and Join DataFrames using Pandas

import pandas as pd

# ----- CONCATENATE OPERATIONS ------------------- (all columns name should be same)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 1:")
print(df1)

#  Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [201, 202, 203, 205],
    "Marks": [85, 78, 80, 65]
}, index=[4, 5, 6, 7])   

print("\nDataFrame 2:")
print(df2)

frames = [df1,df2]
result = pd.concat(frames)
print("concatenate dataframes")
print(result)

# ---------------- MERGE OPERATIONS ---------------- ( at least one column should be same, others can be different  )

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Sakshi", "Rahul", "Ankit", "Priya"]
})

print("DataFrame 1:")
print(df1)

# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 2:")
print(df2)

# Inner Merge
inner_merge = pd.merge(df1, df2, on="ID", how="inner")

print("\nInner Merge:")
print(inner_merge)

# Left Merge
left_merge = pd.merge(df1, df2, on="ID", how="left")

print("\nLeft Merge:")
print(left_merge)

# Right Merge
right_merge = pd.merge(df1, df2, on="ID", how="right")

print("\nRight Merge:")
print(right_merge)

# Outer Merge
outer_merge = pd.merge(df1, df2, on="ID", how="outer")

print("\nOuter Merge:")
print(outer_merge)

# ---------------- JOIN OPERATIONS ---------------- ( columns names in both datframes is different )

# Creating DataFrames for Join
df3 = pd.DataFrame({
    "Name": ["Amit", "Riya", "Karan"]
}, index=[1, 2, 3])

df4 = pd.DataFrame({
    "Salary": [30000, 40000, 50000]
}, index=[1, 2, 3])

print("\nDataFrame 3:")
print(df3)

print("\nDataFrame 4:")
print(df4)

# Joining DataFrames
joined_df = df3.join(df4)

print("\nJoined DataFrame:")
print(joined_df)

# Joining with Different Join Types
left_join = df3.join(df4, how="left")

print("\nLeft Join:")
print(left_join)

right_join = df3.join(df4, how="right")

print("\nRight Join:")
print(right_join)

outer_join = df3.join(df4, how="outer")

print("\nOuter Join:")
print(outer_join)
```

```python
# PRACTICAL 18
# Basic Data Cleaning and Preprocessing

import pandas as pd
import numpy as np

# Loading CSV File
file_path = r"C:\\Users\\LAKSHYA\\Desktop\\weekly task\students.csv"

# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display original dataset
print("========== ORIGINAL DATASET ==========")
print(df)

# 3. Display dataset information
print("\n========== ORIGINAL INFORMATION ==========")
df.info()

# 4. Check missing values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# 5. Fill missing Age values with mean
if "Age" in df.columns:
    df["Age"] = df["Age"].fillna(df["Age"].mean())

# 6. Fill missing Marks values with mean
if "Marks" in df.columns:
    df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# 7. Fill missing City values with "Unknown"
if "City" in df.columns:
    df["City"] = df["City"].fillna("Unknown")

# 8. Remove duplicate records
df = df.drop_duplicates()

print("\n========== AFTER REMOVING DUPLICATES ==========")
print(df)

# 9. Remove extra spaces from Name column
if "Name" in df.columns:
    df["Name"] = df["Name"].str.strip()

# 10. Convert Name column to lowercase
if "Name" in df.columns:
    df["Name"] = df["Name"].str.lower()

# 11. Convert City column to lowercase
if "City" in df.columns:
    df["City"] = df["City"].str.lower()

# 12. Convert Age into integer
if "Age" in df.columns:
    df["Age"] = df["Age"].astype(int)

# 13. Rename Marks column as Score
if "Marks" in df.columns:
    df.rename(columns={"Marks": "Score"}, inplace=True)

# 14. Sort dataset by Score
if "Score" in df.columns:
    df = df.sort_values(by="Score", ascending=False)

# 15. Reset the index
df = df.reset_index(drop=True)

# 16. Display cleaned dataset
print("\n========== CLEANED DATASET ==========")
print(df)

# 17. Check missing values after cleaning
print("\n========== MISSING VALUES AFTER CLEANING ==========")
print(df.isnull().sum())

# 18. Display final dataset information
print("\n========== FINAL DATASET INFORMATION ==========")
df.info()
```

```python
# PRACTICAL 19
# Calculate Basic Statistics from a Dataset

import pandas as pd
import numpy as np

# Loading CSV File
file_path = r"C:\\Users\\LAKSHYA\\Desktop\\weekly task\students.csv"


# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display the dataset
print("========== DATASET ==========")
print(df)

# 3. Select Marks column
marks = df["Marks"]

# 4. Remove missing values
marks = marks.dropna()

# 5. Count number of observations
count = marks.count()

# 6. Calculate total marks
total = marks.sum()

# 7. Calculate Mean
mean = marks.mean()

# 8. Calculate Median
median = marks.median()

# 9. Calculate Mode
mode = marks.mode()

# 10. Calculate Standard Deviation
std = marks.std()

# 11. Calculate Variance
variance = marks.var()

# 12. Find Minimum
minimum = marks.min()

# 13. Find Maximum
maximum = marks.max()

# 14. Calculate Range
data_range = maximum - minimum

# 15. Calculate First Quartile
q1 = marks.quantile(0.25)

# 16. Calculate Third Quartile
q3 = marks.quantile(0.75)

# 17. Display results
print("\n========== BASIC STATISTICS ==========")

print("Number of Observations :", count)
print("Total Marks            :", total)
print("Mean                   :", mean)
print("Median                 :", median)
print("Mode                   :", mode.tolist())
print("Standard Deviation     :", std)
print("Variance               :", variance)
print("Minimum                :", minimum)
print("Maximum                :", maximum)
print("Range                  :", data_range)
print("First Quartile (Q1)    :", q1)
print("Third Quartile (Q3)    :", q3)
```

```python
# Data Visualization using Seaborn

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Creating Sample Dataset
data = {
    "Maths": [85, 90, 78, 92, 88],
    "Science": [80, 85, 75, 95, 89],
    "English": [78, 83, 72, 90, 84]
}

# Creating DataFrame
df = pd.DataFrame(data)

print("Dataset:")
print(df)

# -------------------- HEATMAP --------------------
sns.heatmap(df, annot=True, cmap="coolwarm")

plt.title("Student Marks Heatmap")
plt.show()

#------------BOX PLOT-------------------------

plt.figure(figsize=(5,4))
sns.boxplot(data=df)
plt.title("Box plot")
plt.show()

#----------------PAIRPLOT--------------------

sns.pairplot(df)
plt.show()
```
