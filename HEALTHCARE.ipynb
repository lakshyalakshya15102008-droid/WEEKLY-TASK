```python
#Text Standardization
df["Gender"] = df["Gender"].str.title()
df["Medical Condition"] = df["Medical Condition"].str.title()
df["Admission Type"] = df["Admission Type"].str.title()
df["Insurance Provider"] = df["Insurance Provider"].str.title()
```

```python
#Condition filtering
#Female patients with Diabetes
df[(df["Gender"] == "Female") & (df["Medical Condition"] == "Diabetes")]

#Age < 30 and Elective admission
df[(df["Age"] < 30) & (df["Admission Type"] == "Elective")]

#High Billing Amount > 40000 and Male
df[(df["Billing Amount"] > 40000) & (df["Gender"] == "Male")]

#Cancer patients with A+ Blood Type
df[(df["Medical Condition"] == "Cancer") & (df["Blood Type"] == "A+")]

#3 conditions - Age > 50, Female, Insurance = Medicare
df[
    (df["Age"] > 50) & 
    (df["Gender"] == "Female") & 
    (df["Insurance Provider"] == "Medicare")
]

```

```python
# Create a new feauture-length of stay
df["Length of Stay"]=(
    pd.to_datetime(df["Discharge Date"])-
    pd.to_datetime(df["Date of Admission"])
).dt.days

df[["Date of Admission","Discharge Date","Length of Stay"]].head()
```

```python
# Create a new feauture-Admission Year
df["Admission Year"]=pd.to_datetime(
   df["Date of Admission"]
).dt.year

df[["Date of Admission","Admission Year"]].head()
```

```python
#Average billing amount by admission year
df.groupby("Admission Year")["Billing Amount"].mean()
```

```python
# Create a new feauture-Admission month
df["Admission Month"]=pd.to_datetime(
   df["Date of Admission"]
).dt.month

df[["Date of Admission","Admission Month"]].head()
```

```python
#Prepare monthly admission date
monthly_admissions=(
    df.groupby("Admission Month")["Name"]
    .count()
    .reset_index()
)
monthly_admissions
```
